### Step 1: Create a sample CDC DataFrame

```python
cdc_data = [
    ('ORD100958','Placed','I'),
    ('ORD100208','Delivered','U'),
    ('TEST001',None,'D')
]

cdc_df = spark.createDataFrame(
    cdc_data,
    ['order_id','order_status','opertaion']
)

cdc_df.show()
```

### Step 2: Separate the changes

### Insert records

```python
from pyspark.sql import functions as F
insert_df = cdc_df.filter(
    F.col("opertaion") == 'I'
)
```

### Update records

```python
update_df = cdc_df.filter(
    F.col('opertaion') == 'U'
)
```

### Delete records

```python
delete_df = cdc_df.filter(
    F.col('opertaion') == 'D'
)
```
