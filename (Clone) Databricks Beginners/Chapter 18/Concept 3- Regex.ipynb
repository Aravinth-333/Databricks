```python
employee=spark.read.format("csv").option("header","true").option("inferSchema","true").load("/Volumes/workspace/employee_project/employee_files/employees_data_quality_test.csv")
```

```python
employee.select("employee_id").show()
```

```python
from pyspark.sql.functions import col
employee_pattern=r"^EMP[0-9]{6}$"
employee.select(
    "employee_id",
   col("employee_id").rlike(employee_pattern).alias("is_valid")
).show()
```

### Finding Valid Employee id Details

```python
valid_employee_id=employee.filter(col("employee_id").rlike
                                 (employee_pattern))
valid_employee_id.show(truncate=False)
```

### Finding Invalid Employee id details

```python
invalid_employee_id=employee.filter(~col("employee_id").rlike(employee_pattern))
invalid_employee_id.show()
```

```python
print("Valid Employee id:",valid_employee_id.count())
print("Invalid Employee id:",invalid_employee_id.count())
```

```python
email_pattern=r"^[A-Za-z0-9.]+@[A-Za-z]+\.[A-Za-z]{3}$"

employee.select(
    "email",
    col("email").rlike(email_pattern).alias("is_valid")
).show(truncate=False)
```

### Finding Valid Email

```python
valid_employee_email=employee.filter(col("email").rlike(email_pattern))
valid_employee_email.show(truncate=False)
```

### Finding Invalid Email

```python
invalid_employee_email=employee.filter(~col("email").rlike(email_pattern))
invalid_employee_email.show(truncate=False)
```

```python
print("Valid Employee Email:",valid_employee_email.count())
print("Invalid Employee Email:",invalid_employee_email.count())
```

```python
employee_validated=employee.withColumn("Valid_employee_id",col("employee_id").rlike(employee_pattern)).withColumn("Valid_employee_email",col("email").rlike(email_pattern))
employee_validated.show(truncate=False)
```
