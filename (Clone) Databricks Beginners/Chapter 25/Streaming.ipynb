### Step 1: Create a streaming DataFrame

```python
from pyspark.sql.functions import col

orders_stream_df = (
    spark.readStream
    .format("rate")
    .option("rowsPerSecond",5)
    .load()
)
```

### Step 2: Display the incoming records

```python
display(
    orders_stream_df,
    checkpointLocation = '/Volumes/workspace/default/my_files/checkpoint_rate_demo'
)
```

### Step 3: Apply a transformation

```python
from pyspark.sql.functions import col

processed_stream_df = orders_stream_df.select(
    col("timestamp"),
    col("value").alias("order_number")
)
```

### Step 4: Display the transformed stream

```python
display(
    processed_stream_df,
    checkpointLocation = '/Volumes/workspace/default/my_files/checkpoint_rate_demo_processed'
)
```
