## 3. Employee Distribution by Department

### Business Question

How many employees are currently associated with
each department?

This helps the HR team understand the size of each
department and identify where the workforce is concentrated.

```python
display(
    employee_df
    .groupBy("department")
    .count()
    .orderBy("count", ascending=False)
)
```

## 4. Average Salary by Department

### Business Question

What is the average salary of employees in each department?

This analysis can help HR compare salary levels across
different areas of the organization.

```python
from pyspark.sql.functions import avg

display(
    employee_df
    .groupBy("department")
    .agg(
        avg("salary").alias("average_salary")
    )
    .orderBy("average_salary", ascending=False)
)
```

### Observation

The results provide an initial comparison of average
salary across departments.

The salary values will be investigated further when
we perform data-quality validation.

## 8. Data Quality Investigation

The raw employee dataset may contain incorrect or
incomplete records.

Before building the cleaning process, we investigate
potential problems such as:

- duplicate employee IDs
- missing required fields
- invalid salaries
- unexpected employee statuses
- invalid manager relationships

At this stage, we identify the problems rather than
modifying the source data.

```python
duplicate_employee_ids_df = (
    employee_df
    .groupBy("employee_id")
    .count()
    .filter("count > 1")
)

display(duplicate_employee_ids_df)
```
