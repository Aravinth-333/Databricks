### Step 1: Load the Silver Orders dataset

```python
silver_df = spark.table(
    "workspace.ecommerce_project.silver_orders"
)
```

### Step 2: Run an analytical query

```python
from pyspark.sql.functions import sum,count

state_summary_df = (
    silver_df
    .groupBy("shipping_state")
    .agg(
        sum("final_amount").alias("total_revenue"),
        count("order_id").alias("total_orders")
    )
)

display(state_summary_df.limit(10))
```

### Step 3: Inspect the query execution plan

```python
state_summary_df.explain(mode="formatted")
```
