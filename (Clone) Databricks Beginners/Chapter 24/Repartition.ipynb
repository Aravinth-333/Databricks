### Step 1: Load the existing Silver dataset

```python
silver_df = spark.table(
    "workspace.ecommerce_project.silver_orders"
)

display(silver_df.limit(10))
```

### Step 2: Check the current number of partitions

```python
from pyspark.sql.functions import spark_partition_id

print(
    "Populated partitions: ",
    silver_df
    .select(spark_partition_id().alias("partition_id"))
    .distinct()
    .count()
)
```

### Step 3: Repartition the DataFrame

```python
repartitioned_df = silver_df.repartition(8)
```

### Step 4: Check the partition distribution after repartitioning

```python
print(
    "Populated partition: ",
    repartitioned_df
    .select(spark_partition_id().alias("partition_id"))
    .distinct()
    .count()
)
```

### Step 5: Repartition using a column

```python
state_repartitioned_df = silver_df.repartition(
    8,
    "shipping_state"
)

print(
    "Populated partion after state repartition:",
    state_repartitioned_df
    .select(spark_partition_id().alias("partition_id"))
    .distinct()
    .count()
)
```
