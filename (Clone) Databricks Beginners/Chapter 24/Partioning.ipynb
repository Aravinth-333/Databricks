### Step 1: Load the existing Silver dataset

```python
silver_df = spark.table(
    "workspace.ecommerce_project.silver_orders"
)

display(silver_df.limit(10))
```

### Step 2: Inspect the dataset

```python
silver_df.printSchema()
```

```python
print('Total rows:      ',silver_df.count())
print("Total columns:   ",len(silver_df.columns))
```

### Step 3: Prepare the order date

```python
from pyspark.sql.functions import to_date

silver_prepared_df = silver_df.withColumn(
    "order_date",
    to_date("order_date")
)

display(silver_prepared_df.select("order_id","order_date").limit(10))
```

### Step 4: Create a partitioned Delta table

```python
(
    silver_prepared_df.write
    .format("delta")
    .mode("overwrite")
    .partitionBy("order_date")
    .saveAsTable("workspace.ecommerce_project.silver_orders_partitioned")
)
```

### Step 5: Query orders for a specific date

```python
%sql
select order_id, order_date, final_amount
from workspace.ecommerce_project.silver_orders_partitioned
where order_date = DATE '2025-01-01';
```

### Step 6: Inspect the query execution plan

```python
%sql
EXPLAIN FORMATTED

select order_id,order_date,final_amount
from workspace.ecommerce_project.silver_orders_partitioned
where order_date = DATE '2025-01-01';
```

### Step 7: Verify the partitioned table

```python
partitioned_df = spark.table(
    "workspace.ecommerce_project.silver_orders_partitioned"
)

print("Partitioned table rows:",partitioned_df.count())

display(
    partitioned_df
    .select(
        "order_id",
        "order_date",
        "final_amount"
    )
    .limit(10)
)
```
