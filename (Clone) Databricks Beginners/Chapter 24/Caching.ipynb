### Step 1: Load the Silver dataset

```python
silver_df = spark.table(
    "workspace.ecommerce_project.silver_orders"
)

display(silver_df.limit(10))
```

### Step 2: Run the first analysis

```python
from pyspark.sql.functions import sum,count

silver_df.agg(
    sum("final_amount").alias("total_sales"),
    count("order_id").alias("order_count")
).show()
```

### Step 3: Run another analysis

```python
silver_df.groupBy("shipping_state").count().show()
```

### Step 4: Understand caching in a standard PySpark environment

```python
cached_df = silver_df.cache()

cached_df.count()

cached_df.groupBy("shipping_state").count().show()
cached_df.unpersist()
```
