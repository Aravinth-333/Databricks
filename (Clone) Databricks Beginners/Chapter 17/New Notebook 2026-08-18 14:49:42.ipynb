```python
from pyspark.sql.types import StructField,StructType,StringType,DoubleType

schema=StructType([
    StructField("emp_id",StringType(),True),
    StructField("first_name",StringType(),True),
    StructField("last_name",StringType(),True),
    StructField("department",StringType(),True),
    StructField("designation",StringType(),True),
    StructField("manager_id",StringType(),True),
    StructField("salary",DoubleType(),True),
    StructField("joining_date",StringType(),True),
    StructField("city",StringType(),True),
    StructField("employee_status",StringType(),True)
])
```

```python
df=spark.read.csv("/Volumes/workspace/employee_project/employee_files/employees.csv",header=True,schema=schema)
df.show(n=5)
```

```python
df.printSchema()
```

```python
orders=spark.read.format("json").option("multiLine",True).load("/Volumes/workspace/employee_project/employee_files/nested.json")
```

```python
orders.show()
```

```python
from pyspark.sql.functions import concat,lit
payment_method=orders.select(concat("payment.method",lit("-"),"payment.status")).withColumnRenamed("concat(payment.method, -, payment.status)","Payment_status")
payment_method.show()
```

```python
# orders.show()
product_table = orders.select("product.product_id","product.quantity")
```

```python
product_table.show()
```

```python
full_table=orders.select(
    "order_id",
    "customer.customer_id",
    "product.product_id",
    "customer.name",
    "product.quantity",
    concat("payment.method",lit("-"),"payment.status").alias("Payment_status")
)
```

```python
full_table.show(truncate=False)
```

Converting json into Parquet

```python
df=spark.read.json("/Workspace/Databricks Beginners/Data/products.json",multiLine=True)
```

```python
df.write.mode("overwrite").parquet("/Volumes/workspace/employee_project/employee_files/product.parquet")
```

```python
product=spark.read.parquet("/Volumes/workspace/employee_project/employee_files/product.parquet")
product.show()
```

```python
df.write.mode("overwrite").format("delta").save("/Volumes/workspace/employee_project/employee_files/product_delta")
```

```python
product=spark.read.format("delta").load("/Volumes/workspace/employee_project/employee_files/product_delta")
product.show()
```

```python
df.write.mode("overwrite").format("delta").saveAsTable("employee_project.product_table")
```
